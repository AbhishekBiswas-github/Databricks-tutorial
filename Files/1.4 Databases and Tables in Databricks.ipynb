# Simple Delta tables

```sql
USE CATALOG demoworkspace_new;

CREATE TABLE managed_default
    (width INT, length INT, height INT);

INSERT INTO managed_default
VALUES(3 INT, 2 INT, 1 INT);
```

### Describing managed table

```sql
DESCRIBE EXTENDED managed_default;
```

### Creating external table

```sql
CREATE TABLE external_default
    (width INT, length INT, height INT)
LOCATION "abfss://unity-catalog-storage@dbstorageb3e5vgewgi4ie.dfs.core.windows.net/7405618761365455/external_default";

INSERT INTO external_default
VALUES(3 INT, 2 INT, 1 INT);
```

### Describing external table

```sql
DESCRIBE EXTENDED external_default;
```

### Dropping Both table

```sql
DROP TABLE managed_default;
DROP TABLE extended_default;
```

# Creation of custom database and its tables 
- Using the custom database 
- creating both managed and external tables 

```sql
USE CATALOG demoworkspace_new;
USE SCHEMA custom;
CREATE TABLE managed_custom
    (width INT, length INT, height INT);

INSERT INTO managed_custom
VALUES(3 INT, 2 INT, 1 INT);
DESCRIBE EXTENDED managed_custom;
CREATE TABLE external_custom
    (width INT, length INT, height INT)
LOCATION "abfss://unity-catalog-storage@dbstorageb3e5vgewgi4ie.dfs.core.windows.net/7405618761365455/external_custom";

INSERT INTO external_custom
VALUES(3 INT, 2 INT, 1 INT);
```

## Describing the database and data files

```sql
DESCRIBE DATABASE EXTENDED custom;
```

```sql
LIST "abfss://unity-catalog-storage@dbstorageb3e5vgewgi4ie.dfs.core.windows.net/7405618761365455/custom"
```
