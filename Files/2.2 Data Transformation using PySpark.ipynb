```python
employees_df = spark.read.json("/Volumes/demoworkspace_new/default/my_volume/json/employees.json")

display(employees_df)
```

```python
from pyspark.sql.functions import col

display(employees_df.select(col("id"), col("name"), col("address"), col("role")))
```

```python
from pyspark.sql.functions import col, split

display(employees_df.select(col("id"), col("name"), col("address.city").alias("city"), col("address.country").alias("country")))
```

```python
from pyspark.sql.functions import col
new_employees_df = employees_df.withColumn("Employee_Id", col("id")) \
                            .withColumn("Employee_Name", col("name"))\
                            .withColumn("City", col("address.city")) \
                            .withColumn("Country", col("address.country"))\
                            .withColumn("Employee_Role", col("role"))\
                            .drop("id", "name", "address", "dept_id", "role")
                            
display(new_employees_df)
```

```python
%sql
SELECT COUNT(id) as `Total Employees`, SUM(salary) as `Total Salary by role`,  AVG(salary) as `Average Salary by role`, role FROM 
json.`/Volumes/demoworkspace_new/default/my_volume/json/employees.json`
GROUP BY role
ORDER BY `Total Employees` DESC;
```
